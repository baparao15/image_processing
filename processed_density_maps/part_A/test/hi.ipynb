```python
import h5py
import matplotlib.pyplot as plt

with h5py.File("IMG_10.h5", "r") as f:
    density = f["density"][:]
    image = f["image"][:]

print("Density shape:", density.shape)
print("Image shape:", image.shape)
print("Estimated count:", density.sum())

plt.imshow(image)
plt.axis("off")
plt.show()

plt.imshow(density, cmap="jet")
plt.colorbar()
plt.title("Density Map")
plt.axis("off")
plt.show()
```
